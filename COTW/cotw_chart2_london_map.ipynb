```python
import os
os.chdir("/Users/student/Library/CloudStorage/OneDrive-LondonSchoolofEconomics/1. LSE/Postgraduate/Growth Lab/SarahRashid0.github.io/COTW")
import altair as alt
import numpy as np
import pandas as pd
import requests
import eco_style
alt.theme.enable("light")
```

```python
# ── URLs ──────────────────────────────────────────────────────────────────────
BASE = "https://raw.githubusercontent.com/SarahRashid0/SarahRashid0.github.io/refs/heads/main/website"

boroughs_url   = "https://raw.githubusercontent.com/Anas1108/London_GeoJson_MapVisualization_using_D3/refs/heads/main/london_boroughs.json"
imd_url        = f"{BASE}/material_deprivation_londonLAD.csv"
schools_url    = f"{BASE}/independent_schools_with_coordinates.csv"
```

```python
boroughs = alt.Chart(
    alt.UrlData(boroughs_url, format=alt.DataFormat(type="json", property="features"))
).transform_lookup(
    lookup="properties.code",
    from_=alt.LookupData(data=imd_url, key="Code", fields=["IMD"])
).mark_geoshape(
    stroke="#FFFFFF", strokeWidth=1
).encode(
    color=alt.Color(
        "IMD:Q",
scale=alt.Scale(
    range=["#e8f5fc", "#179fdb", "#0063AF"],
    interpolate="hcl"
),
        legend=alt.Legend(
            title="IMD score (higher = more deprived)",
            orient="bottom", direction="horizontal",
            gradientLength=180
        )
    ),
    tooltip=[
        alt.Tooltip("properties.name:N", title="Borough"),
        alt.Tooltip("IMD:Q", title="IMD score", format=".1f")
    ]
)

schools = alt.Chart(alt.UrlData(schools_url)).transform_filter(
    "datum.Latitude != null && datum.Longitude != null"
).transform_filter(
    "datum.Latitude > 51.28 && datum.Latitude < 51.69"
).transform_filter(
    "datum.Longitude > -0.51 && datum.Longitude < 0.33"
).transform_filter(
    "datum['GOR (name)'] == 'London'"
).mark_circle(
    color="#e6224b", size=18, opacity=0.55,
    stroke="#FFFFFF", strokeWidth=0.5
).encode(
    longitude="Longitude:Q",
    latitude="Latitude:Q",
    tooltip=[
        alt.Tooltip("EstablishmentName:N", title="School"),
        alt.Tooltip("Gender (name):N", title="Gender")
    ]
)

# ── Compose — projection goes here on the combined chart ──────────────────────
chart = (boroughs + schools).project(
    type="mercator", center=[-0.15, 51.495], scale=32000
).properties(
    width=500,
    height=420,
    title=alt.TitleParams(
        text="Independent schools are unevenly distributed across London's boroughs",
        subtitle="Each dot is an independent school. Shading = Index of Multiple Deprivation (darker = more deprived)",
        anchor="start"
    )
)

chart
```
