```python
import os
os.chdir("/Users/student/Library/CloudStorage/OneDrive-LondonSchoolofEconomics/1. LSE/Postgraduate/Growth Lab/SarahRashid0.github.io/COTW")
import altair as alt
import numpy as np
import pandas as pd
import requests
import eco_style
alt.theme.enable("light")
```

```python
url = "https://raw.githubusercontent.com/SarahRashid0/SarahRashid0.github.io/refs/heads/main/website/regression/multiple_regression_predictions.csv"
df = pd.read_csv(url)
```

```python

# ── Scatter points ────────────────────────────────────────────────────────────
points = alt.Chart(df).mark_point(
    filled=True, size=80, opacity=0.7
).encode(
    x=alt.X(
        "gcse_grade5_pct:Q",
        title="Actual GCSE Grade 5+ (%)",
        scale=alt.Scale(domain=[40, 75])
    ),
    y=alt.Y(
        "predicted_gcse:Q",
        title="Predicted GCSE Grade 5+ (%)",
        scale=alt.Scale(domain=[40, 75])
    ),
    color=alt.value("#179fdb"),
    tooltip=[
        alt.Tooltip("Name:N", title="Borough"),
        alt.Tooltip("gcse_grade5_pct:Q", title="Actual (%)", format=".1f"),
        alt.Tooltip("predicted_gcse:Q", title="Predicted (%)", format=".1f")
    ]
)

# ── Perfect fit line ──────────────────────────────────────────────────────────
line_data = pd.DataFrame({"x": [40, 75], "y": [40, 75]})
line = alt.Chart(line_data).mark_line(
    strokeDash=[5, 5],
    strokeWidth=1.5,
    color="#676A86",
    opacity=0.6
).encode(
    x=alt.X("x:Q"),
    y=alt.Y("y:Q")
)

# ── R² annotation ─────────────────────────────────────────────────────────────
annotation_data = pd.DataFrame({
    "x": [41], "y": [73],
    "text": ["R² = 0.504  |  Adjusted R² = 0.408"]
})
annotation = alt.Chart(annotation_data).mark_text(
    align="left", fontSize=11, color="#676A86"
).encode(
    x=alt.X("x:Q"),
    y=alt.Y("y:Q"),
    text="text:N"
)

# ── Compose ───────────────────────────────────────────────────────────────────
chart = (points + line + annotation).properties(
    width=440,
    height=400,
    title=alt.TitleParams(
        text="Actual vs predicted GCSE Grade 5+, London boroughs",
        subtitle="Multiple regression controlling for income, education, house prices, deprivation",
        anchor="start"
    )
)

chart
```
